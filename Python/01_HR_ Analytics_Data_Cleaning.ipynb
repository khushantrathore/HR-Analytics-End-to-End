### 2. Data Cleaning

```python
#check missing values
```

```python
df.isnull().sum()
```

```python
#fill missing values
```

```python
df["YearsWithCurrManager"] =df["YearsWithCurrManager"].fillna(df["YearsWithCurrManager"].median())
```

```python
#verify missing values
```

```python
df.isnull().sum()
```

**Observation:**
- The "YearsWithCurrManager" column contained 57 missing values.
- Missing values were replaced with the median to preserve the dataset without significantly affecting the analysis.

```python
#check duplicate rows
```

```python
df.duplicated().sum()
```

```python
#remove duplicate rows
```

```python
df = df.drop_duplicates()
```

```python
#verify duplicate rows
```

```python
df.duplicated().sum()
```

**Observation:**
Duplicate records were identified and removed to ensure data quality

### Standarize Inconsistent Values

```python
df["BusinessTravel"] = df["BusinessTravel"].replace({"TravelRarely": "Travel_Rarely"})
```

### Verify The Change

```python
df["BusinessTravel"].unique()
```

**Observation:**
- The 'BusinessTravel' column contained inconsistent values ('TravelRarely` and 'Travel_Rarely').
- These values were standardized to 'Travel_Rarely' to ensure consistency.
- After verification, only consistent values remain in the column.
