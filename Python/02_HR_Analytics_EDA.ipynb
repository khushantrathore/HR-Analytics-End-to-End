```python
#This shows the number of employees who stayed and left.
```

```python
#Visualization
```

```python
df["Attrition"].value_counts().plot(kind="bar")

plt.title("Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

***observations:***

- The Attrition variable indicates whether an employee has left the organization or not. 
- "Yes" represents employees who have left the company, while "No" represents employees who are still working in the organization. 
- The distribution helps understand employee retention and attrition levels.

## 1.2 Bivariate Analysis

### 1.2.1 Attrition vs Gender

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Gender', hue='Attrition')

plt.title('Attrition by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

***Observations:***


- This analysis compares attrition among male and female employees.
- It helps determine whether employee turnover differs by gender and provides insights into gender-wise retention patterns within the organization.
- The chart shows that male employees have a higher number of attrition cases than female employees.

### 1.2.2 Attrition vs Age

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', hue='Attrition', bins=10, kde=True)

plt.title('Attrition by Age')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

***Observation:***

- This analysis shows how attrition varies across different age groups.
- It helps identify whether employees of certain age ranges are more likely to leave the organization than others.
- The chart indicates that attrition is comparatively higher among younger employees, while older employees show lower attrition.

### 1.2.3 Attrition vs Department

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Department', hue='Attrition')

plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

***Observations:***

- This analysis compares attrition across different departments.
- It helps identify which departments have higher or lower employee turnover and provides insights into department-wise retention.
- The "Research & Development" records the highest attrition, whereas the "Human Resources' department has the lowest attrition.

### 1.2.4 Attrition vs Job Role

```python
plt.figure(figsize=(12,6))
sns.countplot(data=df, x='JobRole', hue='Attrition')

plt.title('Attrition by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.legend(title='Attrition')
plt.show()
```

***Observations:***

- This analysis examines employee attrition across different job roles. 
- It helps identify the roles with higher turnover and supports understanding workforce stability in each position. 
- The "laboratory Technician" job role has the highest attrition, while the "Manager role" has one of the lowest attrition counts.

### 1.2.5 Attrition vs Monthly Income

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')

plt.title('Attrition vs Monthly Income')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.show()
```

***Observations:***

- This analysis compares the monthly income of employees who left the organization with those who stayed. 
- The box plot shows that employees with Attrition = No have a higher median monthly income than employees with Attrition = Yes.
- Employees who left the organization generally belong to the lower income range, while employees who stayed have higher and more varied monthly incomes.

### 1.2.6 Attrition vs Education Field

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='EducationField', hue='Attrition')

plt.title('Attrition by Education Field')
plt.xlabel('Education Field')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.legend(title='Attrition')
plt.show()
```

***Observations:***

- This analysis compares employee attrition across different education fields.
- It helps identify which educational backgrounds have higher or lower employee turnover.
- The chart shows that employees from the "Life Sciences" education field have the highest attrition, while the "Human Resources" education field has the lowest attrition.

### 1.2.7 Attrition Vs Marital Status

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='MaritalStatus', hue='Attrition')

plt.title('Attrition by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

***OBservations:***

- This analysis compares employee attrition based on marital status.
- It helps understand how employee turnover varies among different marital status groups.
- The chart indicates that "Single" employees have the highest attrition, followed by "Married" employees, while "Divorced" employees have the lowest attrition.

## Overall observations of Bivariate Analysis


- The bivariate analysis explored the relationship between employee attrition and key demographic and organizational factors.
- The analysis indicates that attrition varies across departments, job roles, age groups, monthly income, education fields, and marital status.
- Employees with lower monthly income, younger age groups, and certain job roles showed comparatively higher attrition.
- These findings provide valuable insights into the factors associated with employee turnover and help identify areas where the organization can focus its employee retention strategies.
